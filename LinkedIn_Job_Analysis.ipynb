```python
data = df.copy()
```

```python
print("Duplicate Rows:", data.duplicated().sum())

data.drop_duplicates(inplace=True)

print("Shape after removing duplicates:", data.shape)
```

```python
data["compensation"] = (
    data["compensation"]
        .astype(str)
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False)
)

data["compensation"] = pd.to_numeric(
    data["compensation"],
    errors="coerce"
)

print("\nCompensation datatype:")
print(data["compensation"].dtype)
```

```python
data["job_domain"] = data["job_domain"].fillna("Unknown")
```

```python
data["ben_pack_id"] = data["ben_pack_id"].fillna(0)
```

```python
text_columns = ["title", "description"]

for col in text_columns:
    data[col] = (
        data[col]
        .fillna("")
        .astype(str)
        .str.lower()
        .str.strip()
    )
```

```python
cat_cols = [
    "pay_period",
    "work_type",
    "job_location",
    "job_domain"
]

for col in cat_cols:
    data[col] = (
        data[col]
        .fillna("Unknown")
        .astype(str)
        .str.strip()
    )
```

```python
data["remote_allowed"] = data["remote_allowed"].astype(int)
data["sponsored"] = data["sponsored"].astype(int)
```

```python
num_cols = [
    "applies",
    "views",
    "compensation",
    "company_id",
    "ben_pack_id"
]

for col in num_cols:
    data[col] = pd.to_numeric(
        data[col],
        errors="coerce"
    )
```

```python
data[num_cols] = data[num_cols].fillna(
    data[num_cols].median()
)
```

```python
train_df = data[data["level"].notna()].copy()
predict_df = data[data["level"].isna()].copy()

sample_size = min(20000, len(train_df))

train_sample = train_df.sample(
    n=sample_size,
    random_state=42
)

print("Training sample:", train_sample.shape)
print("Prediction data:", predict_df.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
```

```python

categorical_features = [
    "title",
    "pay_period",
    "work_type",
    "job_location",
    "job_domain"
]

numeric_features = [
    "compensation",
    "applies",
    "views",
    "remote_allowed",
    "sponsored"
]
```

```python
X = train_sample[
    categorical_features + numeric_features
]

y = train_sample["level"]
```

```python
X_predict = predict_df[
    categorical_features + numeric_features
]
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "numeric",
            "passthrough",
            numeric_features
        )
    ]
)

print("X_train:", X_train.shape)
print("X_valid:", X_valid.shape)
print("X_predict:", X_predict.shape)

print("\nPreprocessing pipeline created successfully.")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier",
        RandomForestClassifier(
            n_estimators=100,
            random_state=42,
            n_jobs=-1,
            class_weight="balanced"
        )
    )
])

print("Training model...")

model.fit(X_train, y_train)

print("Model training completed successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

y_pred = model.predict(X_valid)

accuracy = accuracy_score(y_valid, y_pred)

print("=" * 60)
print(f"Validation Accuracy: {accuracy:.4f}")
print("=" * 60)

print("\nClassification Report:\n")
print(classification_report(y_valid, y_pred))

cm = confusion_matrix(y_valid, y_pred)

fig, ax = plt.subplots(figsize=(9, 7))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=model.classes_
)

disp.plot(
    ax=ax,
    xticks_rotation=45,
    values_format="d"
)

plt.title("Level Prediction - Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
predicted_levels = model.predict(X_predict)

print("Number of predictions:", len(predicted_levels))

print("\nPredicted Level Distribution:")
print(pd.Series(predicted_levels).value_counts())
```

```python
missing_mask = data["level"].isna()

data.loc[missing_mask, "level"] = predicted_levels

print("Missing level values after prediction:")
print(data["level"].isna().sum())
```
